```python
df=df.drop(39)  # drop 39 rows
```

```python
df.dropna() # 刪掉有缺值 , 回傳新列表
```

```python
df.fillna({"OrderId":0}, inplace=True) # 把 OrderId 欄裡的缺值換成 0，inplace並且直接修改 df 本身。
```

```python
df.rename(columns={"Order ID":"orderId"}, inplace=True) # rename Order ID to orderId
```

```python
df.groupby("Country")["Price"].sum() # took each of the country price sum 
```
